## *Docker Compose*.

[*Docker Compose*](https://github.com/docker/compose/) es un componente que permite definir la estructura y los servicios de un cluster de contenedores a partir del *Docker Engine*.

* https://docs.docker.com/compose/

### Vesiones de  *Docker Compose*. 

Al igual que *Docker*, *Docker Compose* ha sufrido una evolución muy acelerada la cual en muchos casos conlleva problemas de compatibilidad con el *Docker Engine*. 

Las diversas versiones de *Docker Compose* pueden ser consultadas desde:

https://github.com/docker/compose/releases

### Distribución de *Docker Compose*.

*Docker Compose* es un proyecto de código abierto, por lo que el proyecto ofrece los binarios para las platafomas más comunes e incluso las versiones de *Docker Enterprise*, *Docker Desktop* y *Docker Toolbox* ya contienen a *Docker Compose*.

### *Docker Compose* como *plugin* de *Docker*.

La versión actual de *Docker Compose* (*Compose* v2) es un *plugin* del *CLI* de *Docker* y se invoca como ```docker compose```. La versión 1, un programa en *Python* que se invocaba como ```docker-compose```, ya no recibe mantenimiento.

* En el *Codespace*, la *feature* *Docker-in-Docker* ya instala el *plugin*.
* *Docker Desktop* lo incluye.
* En *GNU/Linux* se instala con el paquete ```docker-compose-plugin``` del repositorio de *Docker* (véase el cuaderno [02](02_introduccion_a_docker.ipynb)).

* La siguiente celda muestra la versión instalada.

````bash
docker compose version
````

## Utilización de docker compose.

El comando ```docker compose``` es el componente encargado de desplegar uno o más servicios compuesto por un cluster de contenedores a partir de las especificaciones descritas en un archivo de nombre ```docker-compose.yaml```.

```docker compose <subcomando> <opciones> <argumentos>```

### Subcomandos de servicios.

* ```build```  construye un servicio a partir de un archivo ```docker-compose.yaml```.
* ```config``` valida un archivo ```docker-compose.yaml```.
* ```up``` despliega los contendores del servicio.
* ```create``` crea un servicio.
* ```pause``` pausa un servicio. 
* ```unpause``` reanuda un sevicio pausado.
* ```start``` inicia un servicio.
* ```stop```  detiene un serrvicio.
* ```restart``` reinicia un servicio.
* ```top``` despliega los procesos que se encuentran corriendo.
* ```port```  permite ligar los puertos expuestos del servicio.
* ```ps``` enlista los contenedores del servicio.
* ```pull``` descarga las imágenes requeridas por el servicio.
* ```push``` sube a un registry las imágenes de un servicio.
* ```run``` permite ejecutar un comando dirigido al servicio.

### Subcomandos de componentes.

* ```down```   detiene y elimina contenedores, redes, imágenes y volúmenes.
* ```images``` enlista las imágenes que conforman el cluster de contenedores.
* ```kill``` apaga de forma drástica algún contenedor.
* ```logs``` despliega las birtácoras de los contenedores.   
* ```rm``` elimina contenedores detenidos.
* ```scale``` permite definir el número de contenedores que utilizará un servicio.

La referencia de estos subcomandos puede ser consultada en:

https://docs.docker.com/compose/reference/overview/

## El lenguaje de los archivos de *Compose*.

Los archivos de *Compose* (```compose.yaml```, o el nombre anterior ```docker-compose.yaml```) utilizan un lenguaje declarativo basado en *YAML* y los diversos componentes de *Docker* para describir un servicio.

La referencia de dicho lenguaje puede ser consultada en:

https://docs.docker.com/compose/compose-file/

### Ejemplo ilustrativo: [src/11/docker-compose.yaml](src/11/docker-compose.yaml).

``` yaml
services:
    database:
        image: mysql
        environment:
            MYSQL_ROOT_PASSWORD: ${MYSQL_ROOT_PASSWORD:?Defina MYSQL_ROOT_PASSWORD en el archivo .env}
    webserver:
        image: nginx:alpine
        ports:
            - 8081:80
        depends_on:
            - cache
            - database
    cache:
        image: redis
```

* La contraseña de *MySQL* no está escrita en el archivo: ```${MYSQL_ROOT_PASSWORD:?...}``` la toma de la variable de entorno o del archivo ```.env``` del mismo directorio, y detiene *Compose* con un mensaje si no está definida.
* La clave ```version``` de los archivos antiguos ya no se usa: *Compose* la ignora.
* El servidor web se publica en el puerto ```8081```, porque el ```8080``` lo sigue usando el contenedor ```nginx``` del cuaderno 05.

* La siguiente celda cambia al directorio del ejemplo y, si todavía no existe, crea el archivo ```.env``` con una contraseña aleatoria para *MySQL*. La contraseña no se muestra, y ```.env``` está en ```.gitignore```.
* Para definirla a mano, copie ```.env.example``` a ```.env``` y escriba el valor.

````bash
cd src/11
if [ ! -f .env ]; then
    umask 077
    printf 'MYSQL_ROOT_PASSWORD=%s\n' "$(openssl rand -hex 16)" > .env
fi
ls -l .env
````

* La siguiente celda valida el archivo sin mostrarlo (```--quiet```): sin esa opción, ```docker compose config``` imprimiría la configuración resuelta, incluida la contraseña.

````bash
docker compose config --quiet && echo "El archivo es válido."
````

* La siguiente celda levanta los tres servicios en segundo plano (```-d```). La primera vez descarga las imágenes de *MySQL*, *Nginx* y *Redis*.

````bash
docker compose up -d
````

* La siguiente celda lista los contenedores del proyecto.

````bash
docker compose ps
````

* La siguiente celda consulta el servidor web, publicado en el puerto ```8081``` del entorno (en *GitHub Codespaces*, pestaña *Ports*).

````bash
sleep 3
curl -s -o /dev/null -w "Código HTTP: %{http_code}\n" http://localhost:8081/
````

* La siguiente celda detiene y elimina los contenedores y la red del proyecto. Con ```-v``` también eliminaría los volúmenes.

````bash
docker compose down
````
