# Preliminar.

## Entorno del curso.

Este curso se practica en *GitHub Codespaces* (el entorno de desarrollo en la nube de *GitHub*), con la configuración de *Dev Container* del directorio `.devcontainer/` del repositorio.

* En la página del repositorio, abra el menú *Code*, pestaña *Codespaces*, y cree un *Codespace* a partir de la rama principal.
* El *Codespace* trae *Python* 3.12, *Jupyter* (con los *kernels* de *Python* y de *Bash*) y *Docker* con los *plugins* `compose` y `buildx`. No hace falta instalar nada en su equipo.
* Abra los cuadernos en el *Codespace*; cada uno indica su *kernel*.
* *Docker* funciona con *Docker-in-Docker*: dentro del *Codespace* se ejecuta un demonio (*daemon*) de *Docker* propio, en un contenedor privilegiado. Las imágenes, los contenedores, los volúmenes y las redes que cree pertenecen a ese demonio y se pierden al eliminar el *Codespace*.
* Los puertos que publiquen los contenedores (por ejemplo, `8080`) se abren dentro del *Codespace*. Para verlos en el navegador, use la pestaña *Ports* (Puertos) de *Visual Studio Code*, que los redirige automáticamente.

Los cuadernos se ejecutan en orden: varios usan imágenes, contenedores o archivos que crearon los anteriores.

## Alternativa: un equipo propio.

Si prefiere trabajar en su equipo, instale:

* *Docker Desktop* (*Windows* o *macOS*) o *Docker Engine* (*GNU/Linux*).
* *Visual Studio Code* con la extensión *Dev Containers*.

Al abrir el repositorio, *Visual Studio Code* ofrece reabrirlo en el contenedor, con el mismo entorno que en *GitHub Codespaces*.

* https://www.docker.com/products/docker-desktop/
* https://docs.docker.com/engine/install/
* https://code.visualstudio.com/docs/devcontainers/containers

## Credenciales.

Ningún ejercicio obligatorio necesita una cuenta. Los ejercicios opcionales de *Docker Hub* del cuaderno [06](06_gestion_de_imagenes.ipynb) usan una cuenta propia, con un *token* de acceso personal en lugar de la contraseña.

Las credenciales **nunca** se escriben en los cuadernos ni en archivos que se suban al repositorio:

* En *GitHub Codespaces*, se guardan como secretos de *Codespaces* (`DOCKERHUB_USER` y `DOCKERHUB_TOKEN`), que llegan al entorno como variables.
* En un equipo propio, se copian de `.env.example` a un archivo `.env`, que *Git* ignora.

##  ADVERTENCIA:

**Utilice imágenes de contenedores confiables, ya que aún cuando los contenedores no escalan privilegios al equipo anfitrión, son capaces de ejecutar código arbitrario dentro de su entorno.**
